```python
order_database = [
    {'name' : '김철수', 'completed': True, 'price': 60000},
    {'name' : '이영희', 'completed': False, 'price': 35000},
    {'name' : '박민수', 'completed': True, 'price': 25000},
    {'name' : '최지영', 'completed': True, 'price': 85000},
    {'name' : '정성민', 'completed': False, 'price': 90000}
]

total_price = 0
order_count = 0

for i in order_database :
    if i['completed'] is True and i['price'] >= 50000 :
            total_price += i['price']
            order_count += 1

print(f'우수 고객의 주문 건수 : {order_count} , 총 결제 금액 : {total_price}')
```

```python
fruit_seetness = [12.5, 14.2, 11.8, 10.5, 15.0, 13.6, 11.2, 14.8, 12.0, 13.1]

sweetness_max = fruit_seetness[0]
sweetness_min = fruit_seetness[0]
sweetness_total = 0

for i in fruit_seetness :
    if i > sweetness_max :
        sweetness_max = i
    elif i < sweetness_min :
        sweetness_min = i

    sweetness_total += i

sweetness_avg = sweetness_total / len(fruit_seetness)

print(f'최고 당도 : {sweetness_max}, 최저 당도 : {sweetness_min}, 평균 당도 : {sweetness_avg:.2f}')
```

```python
import random

random_num = random.randint(1, 50)

count = 0

while True:
    print('1 ~ 50 사이에서 입력하세요 : ', end='')
    num = int(input())
    count += 1

    if num < 1 or num > 50 :
        print('1 ~ 50 사이의 숫자를 입력해주세요')
    elif num < random_num :
        print(f'{num} - 보물은 더 큰 숫자의 방에 있습니다.')
    elif num > random_num:
        print(f'{num} - 보물은 더 작은 숫자의 방에 있습니다.')
    elif num == random_num :
        print(f'{num} - 축하합니다. {count}회 만에 보물을 찾았습니다.')
        break
```

```python
n = 10
answer = 0
if n % 2 == 1:
    for i in range(1, n+1):
        if i % 2 == 1:
            answer += i
else :
    for i in range(1, n+1):
        if i % 2 == 0:
            answer += i**2

print(answer)
```
