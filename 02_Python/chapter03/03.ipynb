```python
menu = {
    'coffee' : 4000,
    'tea' : 4500
}
print(f"keys : {menu.keys()}")
print(f"values : {menu.values()}")

set_a = {1, 2, 3}
set_b = {3, 4, 5}
print(f"교집합 : {set_a & set_b}")
print(f"합집합 : {set_a | set_b}")
print(f"차집합 : {set_a - set_b}")
```
