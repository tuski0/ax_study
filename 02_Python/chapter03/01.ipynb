## 리스트

```python
fruits = ['Apple', 'Orange', 'Banana']
print(fruits[1])
```

```python
fruits[1] = '오렌지'
print(fruits[1])
```

```python
fruits.append("Melon")
print(fruits)
```

```python
fruits_copy = fruits
```

```python
fruits_copy[1] = 'Orange'
```

```python
print(fruits_copy[1])
```

```python
fruits_copy[1] is fruits[1]
print(fruits_copy)
print(fruits)
```

```python
fruits = ['Apple']
print(f" 추가 전 : {id(fruits)}")
fruits = fruits + ['Orange', 'Banana']
print(f" 추가 후 : {id(fruits)}")
```

```python
fruits_extend = ['Apple']
print(f" 추가 전 : {id(fruits_extend)}")
fruits_extend.extend(['Orange', 'Banana'])
print(f" 추가 후 : {id(fruits_extend)}")
```

```python
print(fruits_extend)
print(fruits_extend.sort())
```
