```python
import time

def download_page(name, delay_time):
    print(f'{name} 작업 시작')
    time.sleep(delay_time)
    print('작업 종료')

start = time.perf_counter()

download_page('메인 페이지 데이터', 2)
download_page('상세 페이지 데이터', 2)

end = time.perf_counter()

print('총 소요 시간 : ' , end - start)
```
