```python
import csv

with open('member.csv', 'w', encoding='UTF-8', newline='') as fp :
    writer = csv.writer(fp)
    writer.writerow(['이름', '나이', '전화번호'])
    writer.writerow(['김철수', 20, '010-1000-1000'])
```

```python
with open('member.csv', 'r', encoding='UTF-8') as fp :
    reader = csv.reader(fp)
    members = list(reader)
    print(members)
```

```python
import json

# dict -> json 파일로 저장
user = {
    'name' : '김영희',
    'age' : 20,
    'mobile' : '010-1000-1000'
}

with open('user.json', 'w', encoding='UTF-8') as fp:
    json.dump(user, fp, indent=4, ensure_ascii=False)
```

```python
class Animal:
  def breath(self):
    print('숨을 쉰다')

class Dog(Animal):
  def run(self):
    print('네 발로 뛴다')

kami = Dog()
kami.breath()
kami.run()
```

```python
class Daddy:
  def __init__(self, name):
    self.name = name

class Son(Daddy):
  def __init__(self, name, age):
    super.__init__(name)
    self.age = age

sin = Son('신짱구', 5)
print(sin.name)
print(sin.age)
```

```python
class Daddy:
    def __init__(self, name):
        self.name = name

class Son(Daddy):
    def __init__(self, name, age):
        super().__init__(name)
        self.age = age

son = Son('신짱구', 5)
print(son.name) # 신짱구
print(son.age) # 5
```

```python
class Wallet:
    def __init__(self):
        self.__money = 10000
    
    def get_money(self):
        return self.__money

    def set_money(self, amount):
        if (self.__money - amount) >= 0:
            self.__money = self.__money - amount
        else :
            print('계좌는 음수가 될 수 없습니다.')
```

```python
wall = Wallet()
print(wall.get_money())
wall.set_money(1000)
print(wall.get_money())
```
