```python
orders = [
    ("Cafe Latte", 5000, 2),
    ("Americano", 4500, 1),
    ("Cheese Cake", 6500, 1)
]

print('=' * 35)
print(f"{'메뉴명':<12}{'단가':>4}{'금액':>9}")
print('-' * 35)

total_receipt_amount = 0
for menu, item_price, qty in orders:
    subtotal = item_price * qty
    total_receipt_amount += subtotal

    print(f"{menu:<12}{item_price:>8,}{qty:>4}{subtotal:>9,}")

print(f"{"합계 금액":<12}{total_receipt_amount:20,}원")
```
