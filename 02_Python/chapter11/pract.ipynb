```python
import asyncio
import time

async def download_page_async(title, seconds):
    await asyncio.sleep(seconds)
    print(f'{title} : 지연 시간 {seconds}')

async def main():

    await asyncio.gather(
        download_page_async('메인 페이지 데이터', 2),
        download_page_async('상세 페이지 데이터', 2)
        )


start = time.perf_counter()
await main()
end = time.perf_counter()
print(f'총 지연 시간 : {end - start}')
```

```python
import asyncio
import time

async def fetch_weather(area, seconds):
    print(f'{area} 지역 데이터 수집 시작')
    await asyncio.sleep(seconds)
    print(f'{area} 지역 데이터 수집 완료')

async def main():
    await asyncio.gather(
        fetch_weather('서울', 1.5),
        fetch_weather('뉴욕', 2.5),
        fetch_weather('런던', 1.0)
    )

start = time.perf_counter()
await main()
end = time.perf_counter()
print(f'총 지연 시간 : {end - start}')
```
