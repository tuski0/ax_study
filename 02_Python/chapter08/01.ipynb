```python
class Bus:
    # 속성 정의
    def __init__ (self):
        self.count = 0 # 승객 수, 변수 초기화 
        self.passengers = []

    def ride(self, passenger): # 탑승
        self.count += 1
        self.passengers.append(passenger)
        print('탑승')

    def off(self, passenger):
        if self.count >= 1:
            self.passengers.remove(passenger)
            self.count -= 1
        print('하차')
    def show_count (self):
        print('승객 수 : ', self.count)

    def print_self(self):
        print(id(self))

    def __str__(self):
        return f'승객 수 : {self.count}, 승객 : {self.passengers}'

    def __repr__ (self):
        return 'REPR'

class Passenger :
    # self.name, self.age -> 객체가 생성될때 만들어지는 인스턴스 변수
    # name, age -> 객체를 생성하기 위해 매개변수로 정의된 지역변수
    def __init__ (self, name, age):
        self.name = name
        self.age = age

    def __str__ (self):
        return f'이름 : {self.name}, 나이 : {self.age}'

    def __repr__(self):
        return f'Passenger(name={self.name}, age={self.age})'
    
```

```python
bus = Bus()
p1 = Passenger()
p2 = Passenger()

bus.ride(p1)
bus.ride(p2)

bus.show_count()
```

```python
bus1 = Bus() # 객체, 클래스 -> 객체 (인스턴스)

# . 연산자를 통해 객체의 인스턴스 변수, 메서드에 접근
print('탑승객 수 : ', bus1.count)
bus1.ride()
bus1.off()
bus1.print_self()

```

```python
bus1 = Bus()
print('bus1 주소 : ', id(bus1))
bus1.print_self()
```

```python
bus = Bus()

p1 = Passenger('김철수', 20)
p2 = Passenger('김영희', 18)

bus.ride(p1)
bus.ride(p2)

print(bus)
print(p1)
print(p2)
```

```python
p1 = Passenger('김철수', 20)
print(p1)  # __str__ 호출
print(repr(p1)) # __repr__ 호출
```
