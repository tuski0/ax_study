```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def info(self):
        return f'책 제목 : {self.title}, 저자 : {self.author}'
```

```python
book1 = Book('자몽 살구 클럽', '한로로')
book2 = Book('인간 실격', '다자이 오사무')

print(book1.info())
print(book2.info())
```

```python
class SmartLamp:
    def __init__(self, model):
        self.model = model
        self.is_on = False

    def turn_on(self):
        self.is_on = True
        print(f'{self.model} 조명이 켜졌습니다.')

    def turn_off(self):
        self.is_on = False
        print(f'{self.model} 조명이 꺼졌습니다.')
```

```python
lamp1 = SmartLamp('휘닉스')
lamp1.turn_on()
lamp1.turn_off()
```

```python
class Hero:
    def __init__(self, name, hp=100):
        self.name = name
        self.hp = hp

    def take_damage(self, damage):
        self.hp -= damage
        print(f'남은 체력 : {self.hp}')
        if self.hp < 0 :
            print(f'{self.name} 캐릭터가 쓰러졌습니다.')
```

```python
hero = Hero('톰')

hero.take_damage(60)
hero.take_damage(50)
```

```python
class Phone :
    def call(self):
        print('통화를 시작합니다.')
```

```python
class SmartPhone(Phone):
    def call(self):
        print('영상 통화를 시작합니다.')

    def search_web(self):
        print('인터넷 검색을 시작합니다.')
```

```python
iphone = SmartPhone()

iphone.call()
iphone.search_web()
```

```python
class DoorLock:
    def __init__(self, password):
        self.__password = password

    def get_password(self):
        return self.__password

    def set_password(self, password):
        if len(password) >= 4:
            self.__password = password
            print('정상 변경 되었습니다.')
        else :
            print('경고 비밀번호는 4자리 이상이여야 합니다.')
```

```python
dl = DoorLock('1234')
print(dl.get_password())
dl.set_password('123')
print(dl.get_password())
dl.set_password('098765')
print(dl.get_password())
```

```python
class User:
    def __init__ (self, name):
        self.name = name

    def get_info(self):
            print(f'이름 : {self.name}')

class VIPUser(User) :
    def __init__(self, name, point):
        super().__init__(name)
        self.point = point

    def get_info(self):
        print(f'이름 : {self.name}, 포인트 : {self.point}')
```

```python
user1 = User('김철수')
user1.get_info()

vip1 = VIPUser('홍길동', 100)
vip1.get_info()
```
