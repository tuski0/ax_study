```python
class Animal:
    def __init__(self, kind):
        self.kind = kind # 종

    def move(self):
        print('동물이 움직임')

class Dog(Animal):
    def __init__(self, kind):
        super().__init__(kind) # 부모쪽 인스턴스 변수인 kind에 초기화

    def move(self):
        super().move() # 부모 쪽 객체인 move() 호출
        print(f'{self.kind} 걸음으로 움직임')


class Cat(Animal):
    def __init__(self, kind):
        super().__init__(kind)

    def move(self):
        super().move() # 부모 쪽 객체인 move() 호출
        print(f'{self.kind} 걸음으로 움직임')
```

```python
dog = Dog('강아지')
cat = Cat('고양이')

dog.move()
cat.move()
```

```python
class Human:
    def move(self):
        print('두 발로 직립보행')
```

```python
human = Human()

print('human 은 Animal 클래스로부터 생성 되었는지 : ', isinstance(human, Animal))
print('human은 Human 클래스로부터 생성 되었는지 : ', isinstance(human, Human))

print('dog은 Animal 클래스로부터 생성 되었는지 : ', isinstance(dog, Animal))
```
