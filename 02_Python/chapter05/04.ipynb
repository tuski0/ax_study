```python
def generate_nums():

    print('1번째')
    yield 10

    print('2번째')
    yield 20

    print('3번째')
    yield 30

```

```python
gen = generate_nums()
gen
```

```python
for _ in range(3):
    print(next(gen))
```

```python
gen = (i for i in range(1, 101) if i % 2 == 0)

print(gen) # <generator object <genexpr> at 0x10ab4a670>
```

```python
for num in gen :
    print(num, end=' ')     
```
