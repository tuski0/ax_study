```python
def message_store() :
    storage = ''
    
    def add_message(message):
        nonlocal storage
        storage += message + '\n'

        return storage


    return add_message

my_storage = message_store()
print(my_storage('새로운 알림이 도착했습니다.'))
print(my_storage('출석 체크가 완료되었습니다.'))
```

```python
def deco_border(callback):

    def wrapper() :
        print('='*20)
        result = callback()
        print('='*20)

        return result

    return wrapper

@deco_border
def show_welcome():
    print('환영합니다 스파르타님')

show_welcome()
```

```python
def auto_tracker(callback):

    def wrapper(*args, **kwargs):
        print(f"[{wrapper.__name__}] 함수 실행 시작")
        print(args)
        print(kwargs)

        result = callback(*args, **kwargs)

        print(f'[{callback.__name__}] 함수 실행 종료')

        return result

    return wrapper

@auto_tracker
def add_numbers(a, b) :
    print(a + b)

add_numbers(a = 20, b= 20)
```
