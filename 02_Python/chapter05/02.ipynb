```python
def factorial1(num):
    result = 1
    for i in range(1, num+1):
        result *= i

    return result

factorial1(5)
```

```python
def factorial2(num):
    if (num < 1):
        return 1

    return num * factorial2(num - 1) # 재귀함수

factorial2(5)
```

```python
import time
start = time.time()
result = factorial1(10)

print(result)

end = time.time()

print(end - start)
```

```python
# 핵심 기능과 공통 기능의 분리, 핵심 기능을 원하는 함수로 변경 가능
def time_checker(callback, num):
    start = time.time() # 공통 기능

    result = callback(num) # 핵심 기능 - 팩토리얼 연산 함수 (factorial1, factorial2) 

    end = time.time() # 공통 기능
    print("총 소요 시간: " , end - start)

    return result

print('결과 값 : ', time_checker(factorial2, 10))
```

```python
def time_checker2(callback):

    def wrapper(*args, **kwargs): 
        start = time.time() # 공통 기능

        result = callback(*args, **kwargs) # 핵심 기능 

        end = time.time() # 공통 기능
        print('소요 시간 : ', end - start) # 공통 기능

        return result 

    return wrapper
```

```python


@time_checker2
def factorial2(num):
    if (num < 1):
        return 1
    return num * factorial2(num - 1)
```

```python

factorial2(10)
```

```python
@time_checker2
def add(num1, num2):
    return num1+num2

print(add(10, 20))
```
