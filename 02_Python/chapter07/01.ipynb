```python
class UserIdMismatchError(Exception):
    pass

class UserPassMismatchError(Exception):
    pass

```

```python
def login(user_id, user_pass):
    if user_id != 'user01':
        raise UserIdMismatchError('아이디가 일치하지 않음') # 예외 발생, 코드 실행 흐름 중단

    if user_pass != '1234':
        raise UserPassMismatchError('비밀번호가 일치하지 않음')

    print('로그인 성공')
```
