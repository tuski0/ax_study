```python
song = ['1991', '7rings', '희재']

it = iter(song)
```

```python
print(next(it))
print(next(it))
print(next(it))
print(next(it)) # StopIteration
```

```python
def generate():
    print('첫번째')
    yield '대박'

    print('두번째')
    yield '사건'
    
    print('세번째')
    yield '성공'
```

```python
gen = generate()

print(next(gen))

print(next(gen))

print(next(gen))
```

```python
import sys

gen = (i**2 for i in range(1, 50001))
lis = [i**2 for i in range(1, 50001)]

print(sys.getsizeof(gen))
print(sys.getsizeof(lis))
```
