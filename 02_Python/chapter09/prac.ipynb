```python
def sum_scores():
    final_score = 50
    bonus_scores = [10, 20, 30]
    for score in bonus_scores:
        final_score += score
    return final_score

result = sum_scores()
```

```python
numbers = [1, 2, -3, 4, -5]
answer = 0

for i in range(len(numbers)):
    for j in range(i, len(numbers)):
        if i == j :
            continue
        if numbers[i] * numbers[j] > answer:
            answer = numbers[i] * numbers[j]

print(answer)
```
