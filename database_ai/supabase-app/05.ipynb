```python
class SupabaseEx:
    def __init__(self):
        self.settings = []

    def table(self, table):
        self.settings.append(table)
        return self

    def select(self, fields):
        self.settings.append(fields)
        return self # 생성된 동일 객체

    def eq(self, name, value):
        self.settings.append(f'{name}={value}')
        return self

    def execute(self):
        print(self.settings)
        print('실행..')
        return '처리된 결과'

supabase = SupabaseEx()

response = (
    supabase.table("users")
        .select('*')
        .eq('email', 'user01@test.org')
        .execute()
)
```

## 빌더 / 생성 패턴

```python
class Product:
    # 내부 식별자
    _sentinel = object()

    # 클래스 외부에서 직접 생성은 차단
    def __init__(self, token, name, price, description):
        if token is not Product._sentinel: # 외부에서 직접 생성한 경우
            raise Exception('직접 생성 불가능')
        
        self.__name = name
        self.__price = price
        self.__description = description

    def name(self):
        return self.__name

    def price(self):
        return self.__price

    def description(self):
        return self.__description

    def __repr__(self):
        return f"Product(name='{self.__name}', price='{self.__price}', description = '{self.__description}')"

    # Product 객체를 조립하고 생성하는 클래스
    class Builder:

        def name(self, name):
            self.__name = name
            return self
        
        def price(self, price):
            self.__price = price
            return self

        def description(self, description):
            self.__description = description
            return self


        # 객체 생성
        def build(self):
            return Product(Product._sentinel, self.__name, self.__price, self.__description)
```

```python
product = (
    Product.Builder()
        .name('Iphone')
        .price(10000)
        .description('duo')
        .build()
)

product
```

```python
print(product.name(), product.price(), product.description())
```

```python
from dataclasses import dataclass

# @dataclass (kw_only=True) -> 키워드 매개변수만 사용
@dataclass
class Product:
    name : str
    price : int = 0
    description : str = ""
    
# 위치 인수 사용
product = Product('iphone', 10000, 'duo')
print(product) # Product(name='iphone', price=10000, description='duo')

# 키워드 인수 사용
product = Product(
    name = 'iphone',
    description = 'Duo',
    price = 3290000
)
print(product) # Product(name='iphone', price=3290000, description='Duo')
```
