# Supabase 연동하기

```python
import os
from dotenv import load_dotenv
from supabase import create_client

# .env 파일의 환경변수 로드
load_dotenv()

url = os.environ.get("SUPABASE_URL")
key = os.environ.get("SUPABASE_PUBLISHABLE_KEY")

# supabase 클라이언트 객체 생성
supabase = create_client(url, key)

print(supabase)
```
