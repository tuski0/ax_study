## 회원 CRUD - auth.users, user_details
- 삭제 : deleted_at에 날짜와 시간이 업데이트되면 소프트 삭제
- 조회시에는 deleted_at IS NOT NULL로 조회될 수 있도록 한다.


```python
import os
from dotenv import load_dotenv
from supabase import create_client

# .env 파일의 환경변수 로드
load_dotenv()

url = os.environ.get("SUPABASE_URL")
key = os.environ.get("SUPABASE_PUBLISHABLE_KEY")

# supabase 클라이언트 객체 생성
supabase = create_client(url, key)

print(supabase)
```

```python
class user_service:
    def __init__(self):
        data = supabase.auth.get_user()
        if data: self.user = data.user

    def create(self, cellphone, zipcode, address, address_sub, type = "BUYER"):
        response = (
            supabase.table('user_details')
                .insert({
                    "user_id" : self.user.id,
                    "type" : type,
                    "cellphone" : cellphone,
                    "zipcode" : zipcode,
                    "address" : address,
                    "address_sub" : address_sub
                })
                .execute()
        )
        return response.data
    
    def update(self, cellphone=None, zipcode=None, address=None, address_sub=None, type = None):
        update_data = {}
        if cellphone is not None:
            update_data["cellphone"] = cellphone
        if zipcode is not None:
            update_data["zipcode"] = zipcode
        if address is not None:
            update_data["address"] = address
        if address_sub is not None:
            update_data["address_sub"] = address_sub
        if type is not None:
            update_data["type"] = cellphone
        
        if not update_data:
            return None
        
        response = (
            supabase.table('user_details')
                .update(update_data)
                .eq("user_id", self.user.id)
                .execute()
        )
        return response.data

    def get(self):
        response = (
            supabase.table("user_details")
                .select("*")
                .eq("user_id",self.user.id)
                .execute()
        )
        return response.data
```
