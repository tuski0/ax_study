```typescript

interface Course {
	title: string,
	credit: number,
	isOnline: boolean,
};

function getCourseValue<T, K extends keyof T>(obj: T, key: K): T[K] {
	return obj[key];
}

const crs1: Course = {
	title: '1231',
	credit: 1125125,
	isOnline: true,
};

const title = getCourseValue(crs1, 'title');
const credit = getCourseValue(crs1, 'credit');

console.log(title);
console.log(credit);
```

```typescript
class Book {
	public title : string;
	private price : number;

	constructor(title, price){
		this.title = title,
		this.price = price;
	};

	setPrice(newPrice : number){
			if (newPrice > 0){
				this.price = newPrice;
			}
	}

	printinfo(): string{
			return `제목 : ${this.title}, 가격: ${this.price}`;
	}
}


const book = new Book('자몽살구 클럽', 10000);
console.log(book.printinfo());
book.setPrice(20000);
console.log(book.printinfo());
```

```typescript
class Beverage {
	constructor (public name: string, protected price: number){}

	public getDetails(): void {
		console.log(`이름 : ${this.name}, 가격: ${this.price}`);
	}
}

class Coffee extends Beverage {
	constructor (name: string, price: number, private shotCount: number) {
		super(name, price);
	}

	public printCoffeeInfo(): void {
			console.log(`이름 : ${this.name}, 가격: ${this.price}, 샷 : ${this.shotCount}`);
	}
}

const cfe = new Coffee('아아', 2000, 2);
cfe.printCoffeeInfo();

cfe.getDetails();
```

```typescript
class User {
	constructor (public userName: string, protected point: number){}

	addpoint(amount: number) : void {
		this.point += amount;
		console.log(this.point);
	}
}

class VIPUser  extends User {
	constructor ( userName, point, private bonusRate : number){
		super(userName, point);
		this.bonusRate = bonusRate;
	}

	addpoint(amount: number) : void {
		this.point += ( amount * this.bonusRate);
		console.log(this.point);
	}
}

const us1 = new User('kim', 0);
const vus1 = new VIPUser('lee', 0, 3);
us1.addpoint(10);
vus1.addpoint(10);
us1.addpoint(10);
vus1.addpoint(10);
```

```typescript
abstract class Animal {
	constructor (public name: string){}

	printName(): void {
		console.log(this.name);
	}

	makeSound(): void {}
}

class Dog extends Animal {
	makeSound(): void {
		console.log('멍멍');
	}
}


const gd = new Dog('GD');
gd.printName();
gd.makeSound();
```

```typescript
interface Powerable {
	turnOn(): boolean;
}

class Television implements Powerable {
	turnOn(): boolean {
		console.log('TV 전원 켜짐');
		return true;
	}
}

class AirConditioner implements Powerable {
	turnOn(): boolean {
		console.log('에어컨 켜짐');
		return true;
	}
}

const tv = new Television();
const air = new AirConditioner();

const devices: Powerable[] = [tv, air];

for (let i = 0; i < devices.length; i++){
	devices[i].turnOn();
}
```
