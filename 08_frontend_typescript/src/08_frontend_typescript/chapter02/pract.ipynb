```typescript
function greetUser(name: string, title?: string) {
	if (title === undefined){
		return `${name} Hello`;
	} else {
		return `${name} ${title} Hello`;
	}
}

console.log(greetUser('홍길동', '회원님'));
console.log(greetUser('이영희'));
```

```typescript
function calculateTotalExpenses(...prices: number[]): number {
    let total = 0;
		for (const price of prices){
			total += price;
		}

		return total;
}

console.log(calculateTotalExpenses(5000, 12000, 3000));
console.log(calculateTotalExpenses(25000, 8000));
```

```typescript
type  transformer= (prices: number) => number;

function processPrices(prices: number[], transformer: transformer): number[]{
	const newPrices = [];
	for (const price of prices) {
		const result = transformer(price);
		newPrices.push(result);
	}

	return newPrices;
}
```

```typescript
processPrices([10000, 20000, 30000], x => x + 1000);
```
