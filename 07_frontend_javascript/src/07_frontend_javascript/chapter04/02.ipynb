#### 2.1 var의 3가지 결함
- 호이스팅 (Hoisting)
- 선언 키워드 생략 허용 (암묵적 전역 변수)
- 함수 레벨 스코프 (Function-level Scope)

```javascript
var i = 10000; // window.i or global.i = 100

for (var i = 0; i < 10; i++){ // window.i or global.i = 0 ~ 10
	// 반복 로직
	console.log("for문에 정의된 i :", i );
}

console.log(i); // window.i or global.i 가 출력
```

### 6. 제어문

#### 6.1 조건문 (if, if-else, if-else if - else)

```javascript
let greeting = '';
greeting = greeting || '기본값';

// gretting을 선언만 하고 값을 대입하지 않았기 때문에 undefined 상태 논리연산자가 false라고 판별하는 값을 가진 경우 
// OR 연산자를 사용하면 다음항의 값을 대입할 수 있기 때문에 기본값 설정으로 사용할 수 있다.
console.log(greeting)

let num = 0;
num = num || 10;
// 0 또한 논리 연산자에서는 false로 인식하기 때문에 num = 10이 된 상태
console.log(num) // 10
```

```javascript
let num = 0;

num = num || 10;
// 0 또한 논리 연산자에서는 false로 인식하기 때문에 num = 10이 된 상태
console.log(num) // 10
```

```javascript
// ?? -> nullish 연산자 -> null, undefined 일 때만 판단
let num = 0;
let num2;
num = num ?? 10;
num2 = num2 ?? 10;
console.log(num)
console.log(num2)
```

```javascript
// && : 단락회로평가를 이용한 간단한 조건식 수행

const message = 'hello';

message && console.log(message); // hello
// 첫번째 항이 참이여야 두번째 항이 실행되는 코드이기 때문에
// 첫번째에 값이 존재하는지 파악하는 용도로 사용 가능
```

#### 6.2 선택문 (switch - case)

```javascript
let rank = 2;

switch(rank) {
	case 1:
		console.log('금메달');
		break;
	case 2:
		console.log('은메달');
		break;
	case 3:
		console.log('동메달');
		break;
	default:
		console.log('ㄲㅂ');
}
```

```javascript
/* let grade = 1 ~ 6;
1 - 1학년, 6 - 6학년, 1 ~ 6 사이의 값이 아니면 잘못 입력함
 */

let grade = 3;

switch (grade) {
  case 1:
    console.log("1학년");
    break;
  case 2:
    console.log("2학년");
    break;
  case 3:
    console.log("3학년");
    break;
  case 4:
    console.log("4학년");
    break;
  case 5:
    console.log("5학년");
    break;
  case 6:
    console.log("6학년");
    break;
  default:
    console.log("값을 잘못 입력함");
}
```

#### 6.3 반복문 (for, while, do-while)

for 문 : 정해진 횟수만큼 반복하기

```javascript
num = 0;
for (let i = 1; i <= 100; i++){
    num += i;
}

console.log(num); // 5050
```

```javascript
for (let i = 0; i < 10; i++){
    console.log(i)
}
```

```javascript
// 구구단 프로그램 만들기
for (let i = 2; i <= 9; i++){
    console.log(`===== ${i}단 =====`)
    for (let j = 1; j <= 9; j++){
        console.log(`${i} * ${j} = ${i*j}`)
    }
    console.log('===============')
}
```

```javascript
// 단보다 곱하는 수가 클때만 출력
for (let i = 2; i <= 9; i++){
    console.log(`===== ${i}단 =====`)
    for (let j = 1; j <= 9; j++){
        if(i >= j) continue;
        console.log(`${i} * ${j} = ${i*j}`)
    }
    console.log('===============')
}
```

```javascript
for(let i = 0; i < 10; i++){
    if(i == 3) break;
    console.log(i) 
}
```
