```javascript
function greetUser(a, b) {
    let name = a;
    let age = b;

    return `안녕하세요, ${name}님! ${age}세가 되신 것을 진심으로 축하합니다.`
}

```

```javascript
const test = greetUser('KIM', 20);

console.log(test);
```

```javascript
function performCalculation(num1, num2, callback){
		return callback(num1, num2);
}
```

```javascript
const add = performCalculation(12, 4, (a, b) => a + b);
console.log(add);
```

```javascript
const multiply = performCalculation(12, 4, (a, b) => a * b);
console.log(multiply);
```

```javascript
function createPiggyBank() {
	let balance = 0;

	function deposit(amount){
		balance += amount;
	}

	function getBalance(){
		return balance;
	}

	return {
    deposit: deposit,
    getBalance: getBalance
  };
}

let myPiggyBank = createPiggyBank();
```

```javascript
myPiggyBank.deposit(1500);

myPiggyBank.getBalance();
```
