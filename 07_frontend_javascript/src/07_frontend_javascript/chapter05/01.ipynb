매개변수 기본값 (Default Parameter)

```javascript
function add(num1, num2 = 10){
	return num1 + num2;
}

add(10);
```

arguments 객체

```javascript
// 함수 호출 시 사용한 인수 목록
function sum(){
	let total = 0;
	for (const arg of arguments) {
		total += arg;
	}
	return total;
}

sum(10, 20, 30, 40);
```

나머지 매개변수 (Rest Parameter, 가변 매개변수)

```javascript
// 가변 매개변수
function sum(...params){
	console.log(params);
}
sum(10, 20, 30, 40);
```

```javascript
// 나머지 매개변수
function sum(num1, ...params){
	console.log(num1, params)
}

sum(10, 20, 30, 40);
```

```javascript
// 전개 연산자
function add(num1, num2){
	return num1+num2;
}

const nums = [10, 20];

console.log(add(nums[0], nums[1]));
console.log(add(...nums)); // 배열 내부 값 전개해서 매개변수로 전달
```

```javascript
const nums1 = [10, 20, 30, 40];
// const nums2 = nums1; // 주소 복사
const nums2 = [...nums1]; // nums1의 변수를 전개해서 대입함으로써 값은 동일하지만 주소는 다른 배열 생성
console.log(nums2, nums1 === nums2);
```

```javascript
const person = {name: 'kim', age:20};
const person2 = {...person};
console.log(person2, person === person2);
```

```javascript
const person3 = {...person, address: '주소'};
// ...person을 전개해서 person3에 대입하고, 
// person에 address 값이 존재하는 경우에는 '주소'로 덮어쓰기, 존재하지 않는 경우 신규 생성
console.log(person, person3);
```

#### 4. 값으로서의 함수 (일급 객체)와 콜백

#### 4.1 용어의 유래: 차별 없는 '동등한 권리’

#### 4.2 일급 객체로서의 3대 속성
- 변수 할당 (Assignment)
- 매개변수 전달 (Argument Passing)
- 반환값 활용 (Return Value)

```javascript
function square(x){
	return x * x;
}

function calc(nums, callback){
	let total = 0;
	for(const num of nums){
		if (typeof callback === 'function'){
			total += callback(num);
		}
	}

	return total;
}
```

```javascript
// 이미 선언되어 있는 함수를 함수 호출시에 전달하여 callback 함수로 사용
calc([10, 20, 30], square);
```

```javascript
// 일회용 함수, 함수 호출시에 callback 함수를 선언과 동시에 호출
calc([10, 20, 30], function(num){
	return num* num;
});
```

```javascript
// 일회용으로 사용할 함수를 단축하여 사용하기 위해 화살표 함수를 사용
calc([10, 20, 30], x => x * x);
```

### 5. 화살표 함수 (Arrow Function)

#### 5.1 정의 방법 및 스코프 생략 규칙

- function 키워드 생략
- 매개변수 정의부와 코드 실행 영역({}) 사이에 => 로 표기

```javascript
function square(num){
	return num * num;
}
```

```javascript
const square2 = (num) => {
	return num * num;
};

square2(10);
```

```javascript
// const square3 = (num) => num * num;
// const square3 = (x) => x * x;
const square3 = x => x * x;

square3(10);
```

#### 5.2 화살표 함수의 제한 사항
- 생성자 함수 사용 불가
- arguments 객체 미지원
- yield 사용 불가

#### 5.3 일반 함수 vs 화살표 함수의 this 바인딩 비교

- 일반 함수의 this : 함수가 호출되면 호출한 객체
- 화살표 함수의 this : 함수가 정의되는 시점에 이미 결정되어 있는 this로 정의
