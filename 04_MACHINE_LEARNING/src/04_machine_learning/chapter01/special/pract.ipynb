```python
import pandas as pd

df = pd.read_csv('dataset.csv')
```

```python
print(df.isnull().sum())
```

```python
artists_mode = df['artists'].mode()
df['artists'] = df['artists'].fillna(artists_mode[0])
print(df.isnull().sum())
```

```python
trackname_mode = df['track_name'].mode()
df['track_name'] = df['track_name'].fillna(trackname_mode[0])
print(df.isnull().sum())
```

```python
albumname_mode = df['album_name'].mode()
df['album_name'] = df['album_name'].fillna(albumname_mode[0])
print(df.isnull().sum())
```
