#### 2.3 데이터 변경 및 인덱스 제어(set_index, reset_index)

```python
df['보너스'] = df['급여'] * 0.1
print(df)
#     이름  부서   급여   보너스
# 0  김철수  개발  4500  450.0
# 1  이영희  기획  5200  520.0
# 2  박민수  개발  4800  480.0
```

```python
# 인덱스 특정 컬럽으로 변경하기
df_indexed = df.set_index('이름')
print(df_indexed)
#       부서   급여   보너스
# 이름                  
# 김철수  개발  4500  450.0
# 이영희  기획  5200  520.0
# 박민수  개발  4800  480.0
```

```python
# 인덱스를 다시 숫자형 인덱스로 변경하기
df_restored = df_indexed.reset_index()

print(df_restored)
#     이름  부서   급여   보너스
# 0  김철수  개발  4500  450.0
# 1  이영희  기획  5200  520.0
# 2  박민수  개발  4800  480.0
```

#### 2.4 Pandas Series & DataFrame 브로드캐스팅 연산

```python
df = pd.DataFrame({
    '중간고사' : [80, 95, 70],
    '기말고사' : [85, 90, 75]
}, index=['김철수', '이영희', '박민수']) 

print(df)
#      중간고사  기말고사
# 김철수    80    85
# 이영희    95    90
# 박민수    70    75

```

```python
score_means = df.mean(axis=0) # 행 기준 ( 인덱스 기준 )
print(score_means)

# 중간고사    81.666667
# 기말고사    83.333333
# dtype: float64ㅠ
```

```python
score_gap = df.sub(score_means, axis=1)
print(score_gap)
#          중간고사    기말고사
# 김철수  -1.666667  1.666667
# 이영희  13.333333  6.666667
# 박민수 -11.666667 -8.333333
```

### 3. Pandas 정밀 조회 및 결측치 제어

#### 3.1  loc와 iloc를 활용한 행·열 선택

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "나이" : [25, 30, 35, 40],
    "점수" : [80, 95, 70, 85],
    "등급" : ['B', 'A', 'C', 'D']
}, index=['user1', 'user2', 'user3', 'user4'])
```

```python
print(df.loc['user1'])
# 나이    25
# 점수    80
# 등급     B
# Name: user1, dtype: object
print(df.loc['user1', '등급']) # B

df.loc['user1' : 'user3' , ['점수', '등급']]
# 	   점수	등급
# user1	80	B
# user2	95	A
# user3	70	C
```

```python

df.loc['user1' : 'user3' , ['점수', '등급']]
# 	   점수	등급
# user1	80	B
# user2	95	A
# user3	70	C
```

```python
print(df.loc['user1':'user2', '점수':'등급'])
#        점수 등급
# user1  80  B
# user2  95  A
```

```python
print(df.iloc[0:2, 1:3])
#        점수 등급
# user1  80  B
# user2  95  A₩
```

#### 3.3 결측치(Missing Data) 7대 처리 기법
3.3.1 결측치 탐지 및 현황 요약 (Detection)

```python
import numpy as np
# 결측치 - np.nan, pd.NA, NONE
raw_df = pd.DataFrame({'A': [1, np.nan, 3], 'B': [pd.NA, 5, 6], 'C': [7, 8, 9]})
```

```python
raw_df.isnull().sum()
# A    1
# B    1
# C    0
# dtype: int64
```

3.3.2 결측치 삭제 (Drop)

```python
raw_df.dropna(axis=1)
```

3.3.3 통계적 대푯값 대치 (Imputation)

```python
df_age = pd.DataFrame({'나이': [20, 25, None, 30, 95]})
```

```python
median_val = df_age['나이'].median()
df_age['나이'] = df_age['나이'].fillna(median_val)
print(df_age)

#    나이
# 0  20.0
# 1  25.0
# 2  27.5
# 3  30.0
# 4  95.0
```

3.3.4 시계열 직전/직후 값 대치 (Forward / Backward Fill)

```python
ts_df = pd.DataFrame({'온도': [18.2, None, None, 21.0]})

# ffill
ts_df.ffill()
#   온도
# 0	18.2
# 1	18.2
# 2	18.2
# 3	21.0

# bfill
ts_df.bfill()
#    온도
# 0	18.2
# 1	21.0
# 2	21.0
# 3	21.0
```

3.3.5 선형 보간법 (Linear Interpolation)

```python
speed_df = pd.DataFrame({'속도': [10.0, None, None, 40.0]})

speed_df['속도'] = speed_df['속도'].interpolate(method='linear')
print(speed_df)
#     속도
# 0  10.0
# 1  20.0
# 2  30.0
# 3  40.0
```

3.3.6 이상 기호 치환 및 수치형 변환 (Replace & Type Casting)

```python
survey_df = pd.DataFrame({'점수': [85, -999, '?', 95]})

survey_df.info()
```

```python
survey_df.describe()
```

```python
survey_df['점수'] = survey_df['점수'].replace([-999, '?'], pd.NA)
print(survey_df)
#      점수
# 0    85
# 1   <NA>
# 2   <NA>
# 3    95
```

```python
survey_df['점수'] = survey_df['점수'].replace([-999, '?'], pd.NA)
print(survey_df) # type이 object여서 통계 확인 불가능
#      점수
# 0    85
# 1   <NA>
# 2   <NA>
# 3    95

survey_df['점수'] = pd.to_numeric(survey_df['점수'], errors='coerce')
survey_df.describe() # to_numeric으로 type을 실수로 변경하여 통계 확인 가능

# 점수
# count	2.000000
# mean	90.000000
# std	7.071068
# min	85.000000
# 25%	87.500000
# 50%	90.000000
# 75%	92.500000
# max	95.000000
```

#### 3.4 그룹화 집계(groupby)와 .to_numpy() 상호 변환

```python
sales_df = pd.DataFrame({'지점': ['서울', '서울', '부산', '부산'], '매출': [100, 150, 80, 120]})
```

```python
sales_df = pd.DataFrame({'지점': ['서울', '서울', '부산', '부산'], '매출': [100, 150, 80, 120]})
grouped_df = sales_df.groupby('지점').mean(numeric_only=True) # 숫자 아닌 데이터가 있을 경우 숫자만 골라서 연산함
print(grouped_df)
#       매출
# 지점       
# 부산  100.0
# 서울  125.0

sales_df['매출'].to_numpy(dtype=np.float64)
# array([100., 150.,  80., 120.])
```
